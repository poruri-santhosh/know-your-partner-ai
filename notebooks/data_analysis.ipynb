# Know Your Partner AI: Exploratory Data Analysis

This notebook explores the pairwise compatibility dataset generated from behavioral profiles, examining dimension difference distributions and correlation with compatibility outcome scores.

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load dataset
csv_path = Path('../backend/app/data/datasets/compatibility_data.csv')
df = pd.read_csv(csv_path)
print(f"Dataset shape: {df.shape}")
df.head()
```

```python
# Summary statistics
df.describe()
```

```python
# Correlations with compatibility target score
correlations = df.corr()['compatibility_score'].sort_values()
print("Correlations with Compatibility Score:")
print(correlations)
```
